# Thực hiện yêu cầu 2

*   Hiển thị ảnh tương ứng với kênh Red, Green và Blue
*   Chuyển ảnh màu sang ảnh xám và Hiển thị ảnh xám



Tạo subplot 2x2 để hiển thị 4 ảnh: green, red, blue, gray

```python
fig, axs  = plt.subplots(2, 2, figsize=(10, 10))
# Tách các kênh màu
red_channel = Lena.copy()
red_channel[:, :, 1] = 0
red_channel[:, :, 2] = 0


green_channel = Lena.copy()
green_channel[:, :, 0] = 0
green_channel[:, :, 2] = 0

blue_channel = Lena.copy()
blue_channel[:, :, 0] = 0
blue_channel[:, :, 1] = 0

gray_img = cv2.cvtColor(Lena, cv2.COLOR_RGB2GRAY)

axs[0, 0].imshow(red_channel, cmap='Reds')
axs[0, 0].set_title('Red')

axs[0, 1].imshow(green_channel, cmap='Greens')
axs[0, 1].set_title('Green')

axs[1, 0].imshow(blue_channel, cmap='Blues')
axs[1, 0].set_title('Blue')
axs[1, 1].imshow(gray_img, cmap='gray')
axs[1, 1].set_title('Gray')

plt.show()
```

```python
def display_histogram_RGB(image_path, bins=256):
    # Đọc ảnh
    image = cv2.imread(image_path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # Chuyển từ BGR sang RGB

    # Tạo histogram cho từng kênh màu
    colors = ('r', 'g', 'b')
    plt.figure()
    plt.title('RGB Histogram')
    plt.xlabel('Pixel Intensity')
    plt.ylabel('Frequency')

    for i, color in enumerate(colors):
        hist = cv2.calcHist([image], [i], None, [bins], [0, 256])
        plt.plot(hist, color=color)

    plt.xlim([0, 256])
    plt.show()
```

```python
# Thay thế 'path_to_image' bằng đường dẫn đến ảnh của bạn
image_path =lena_path
display_histogram_RGB(image_path)
```
